````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {176: [['3 6\n1 3 4', '2', True, None], ['2 3\n2 5', '-1', True, None], ['1 10000\n1', '10000', False, None], ['1 9999\n2', '-1', False, None], ['3 10000\n7 11 13', '770', False, None], [None, '3', False, 'import random\nrandom.seed(201)\nn, k = 100, 10000\nprint(n, k)\nprint(*random.sample(range(50, 10001), n))']], 177: [['3\n10 30 5 60', '4500', True, None], ['1\n5 7', '0', False, None], ['2\n1 2 3', '6', False, None], ['4\n5 4 6 2 7', '158', False, None], [None, '48032696', False, 'import random\nrandom.seed(202)\nn = 200\nprint(n)\nprint(*[random.randint(1, 500) for _ in range(n + 1)])']], 178: [['banana', '2', True, None], ['abc', '3', True, None], ['a', '1', False, None], ['aa', '1', False, None], ['ab', '2', False, None], ['racecarannakayak', '3', False, None], [None, '313', False, "import random\nrandom.seed(203)\nprint(''.join(random.choice('ab') for _ in range(2000)))"], [None, '1', False, "import random\nrandom.seed(204)\nprint('a' * 2000)"], [None, '1754', False, "import random\nrandom.seed(205)\nprint(''.join(random.choice('abcdefghijklmnopqrstuvwxyz') for _ in range(2000)))"]], 179: [['4\n0 3 0 7\n2 0 4 0\n0 5 0 1\n6 0 2 0', '14', True, None], ['3\n0 1 0\n0 0 1\n0 0 0', '-1', True, None], ['2\n0 5\n7 0', '12', False, None], ['2\n0 5\n0 0', '-1', False, None], [None, '1442397', False, 'import random\nrandom.seed(206)\nn = 12\nprint(n)\nfor i in range(n):\n    print(*[0 if i == j else random.randint(1, 10**6) for j in range(n)])'], [None, '529', False, 'import random\nrandom.seed(207)\nn = 12\nprint(n)\nfor i in range(n):\n    print(*[0 if i == j or random.random() < 0.6 else random.randint(1, 100) for j in range(n)])']], 180: [['5\n1 2 3 4 5\n1 2\n1 3\n3 4\n3 5', '11', True, None], ['1\n7', '7', False, None], ['2\n5 9\n1 2', '9', False, None], ['3\n5 1 5\n1 2\n2 3', '10', False, None], [None, '666053829', False, 'import random\nrandom.seed(208)\nn = 200000\nprint(n)\nprint(*[random.randint(1, 10**4) for _ in range(n)])\nfor v in range(2, n + 1):\n    print(random.randint(1, v - 1), v)'], [None, '586915409', False, 'import random\nrandom.seed(209)\nn = 200000\nprint(n)\nprint(*[random.randint(1, 10**4) for _ in range(n)])\nfor v in range(2, n + 1):\n    print(v - 1, v)']]}
TL = {176: 3, 177: 3, 178: 3, 179: 4, 180: 3}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
