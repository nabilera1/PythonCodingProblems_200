**사용법**
1. 바로 아래 **🔧 채점기 준비** 셀을 먼저 실행합니다(▶).
2. 각 문제의 `%%writefile` 셀에 코드를 작성하고 실행하면 파일로 저장됩니다.
3. 그 아래 `check(번호)` 셀을 실행하면 채점됩니다.
4. 직접 입력을 넣어 실행해 보려면 새 셀에 `%run p001.py`처럼 입력합니다.

> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {21: [['3 7 5', '7', True], ['-1 -5 -3', '-1', True], ['4 4 2', '4', False], ['2 2 2', '2', False], ['1 2 3', '3', False], ['3 2 1', '3', False], ['-1000000000 0 1000000000', '1000000000', False]], 22: [['3 5', '1', True], ['0 -2', '축 위', True], ['-3 5', '2', False], ['-3 -5', '3', False], ['3 -5', '4', False], ['0 0', '축 위', False], ['7 0', '축 위', False], ['-1000 1', '2', False]], 23: [['가위\n보', 'A 승', True], ['바위\n바위', '무승부', True], ['가위\n바위', 'B 승', False], ['보\n가위', 'B 승', False], ['보\n바위', 'A 승', False], ['바위\n보', 'B 승', False], ['가위\n가위', '무승부', False], ['바위\n가위', 'A 승', False]], 24: [['130 9 Y', '탑승 가능', True], ['119 15 Y', '탑승 불가', True], ['130 9 N', '탑승 불가', False], ['120 10 N', '탑승 가능', False], ['150 30 Y', '탑승 가능', False], ['100 5 N', '탑승 불가', False], ['125 12 N', '탑승 가능', False]], 25: [['12 3', '약수', True], ['12 0', '약수 아님', True], ['12 5', '약수 아님', False], ['0 7', '약수', False], ['-12 4', '약수', False], ['7 7', '약수', False], ['0 0', '약수 아님', False], ['1000000007 1', '약수', False]]}
import subprocess, sys, pathlib

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def check(pid, time_limit=2.0):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    ok = 0
    for i, (inp, exp, shown) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        try:
            r = subprocess.run([sys.executable, str(path)], input=inp + "\n",
                               capture_output=True, text=True, timeout=time_limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({time_limit}초)"); continue
        got = _norm(r.stdout)
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
