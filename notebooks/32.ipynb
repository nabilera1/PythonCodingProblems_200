````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {156: [['6\n1 4\n3 5\n0 6\n5 7\n3 9\n7 10', '3', True, None], ['1\n0 1', '1', False, None], ['3\n1 2\n2 3\n3 4', '3', False, None], ['3\n1 10\n2 3\n4 5', '2', False, None], [None, '11360', False, 'import random\nrandom.seed(109)\nn = 200000\nprint(n)\nfor _ in range(n):\n    s = random.randint(0, 10**9 - 1)\n    print(s, s + random.randint(1, 10**6))']], 157: [['4\n1 4\n2 5\n4 6\n5 7', '2', True, None], ['1\n3 4', '1', False, None], ['3\n1 2\n2 3\n3 4', '1', False, None], ['3\n1 9\n1 9\n1 9', '3', False, None], [None, '1126', False, 'import random\nrandom.seed(110)\nn = 200000\nprint(n)\nfor _ in range(n):\n    s = random.randint(0, 10**7)\n    print(s, s + random.randint(1, 10**5))']], 158: [['4\n2 3 1\n5 2 4 1', '18', True, None], ['2\n7\n3 1000000000', '21', False, None], ['5\n1 1 1 1\n5 4 3 2 1', '14', False, None], ['5\n1 1 1 1\n1 2 3 4 5', '4', False, None], [None, '4791908805445359199', False, 'import random\nrandom.seed(111)\nn = 200000\nprint(n)\nprint(*[random.randint(1, 10**9) for _ in range(n - 1)])\nprint(*[random.randint(1, 10**9) for _ in range(n)])']], 159: [['5\n2 50\n1 10\n2 20\n1 30\n3 5', '85', True, None], ['1\n1 7', '7', False, None], ['3\n1 1\n1 2\n1 3', '3', False, None], ['3\n5 1\n5 2\n5 3', '6', False, None], [None, '43741235494406', False, 'import random\nrandom.seed(112)\nn = 200000\nprint(n)\nfor _ in range(n):\n    print(random.randint(1, 50000), random.randint(1, 10**9))']], 160: [['4\n1 3\n2 5\n6 8\n7 9', '7', True, None], ['1\n-1000000000 1000000000', '2000000000', False, None], ['3\n1 2\n2 3\n3 4', '3', False, None], ['3\n1 10\n2 3\n4 5', '9', False, None], [None, '787583785', False, 'import random\nrandom.seed(113)\nn = 200000\nprint(n)\nfor _ in range(n):\n    a = random.randint(-10**9, 10**9 - 1)\n    print(a, min(10**9, a + random.randint(1, 10**4)))']]}
TL = {156: 2, 157: 2, 158: 2, 159: 2, 160: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
