3. 그 아래 `check(번호)` 셀을 실행하면 채점됩니다.
4. 직접 입력을 넣어 실행해 보려면 새 셀에 `%run p001.py`처럼 입력합니다.

> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {96: [['123\n132', '1S 2B', True, None], ['123\n456', 'OUT', True, None], ['123\n123', '3S 0B', False, None], ['123\n312', '0S 3B', False, None], ['987\n789', '1S 2B', False, None], ['159\n951', '1S 2B', False, None], ['246\n248', '2S 0B', False, None]], 97: [['1994', 'MCMXCIV', True, None], ['4', 'IV', True, None], ['1', 'I', False, None], ['9', 'IX', False, None], ['14', 'XIV', False, None], ['40', 'XL', False, None], ['90', 'XC', False, None], ['400', 'CD', False, None], ['2026', 'MMXXVI', False, None], ['3888', 'MMMDCCCLXXXVIII', False, None], ['3999', 'MMMCMXCIX', False, None]], 98: [['MCMXCIV', '1994', True, None], ['III', '3', True, None], ['I', '1', False, None], ['IV', '4', False, None], ['IX', '9', False, None], ['XLII', '42', False, None], ['MMXXVI', '2026', False, None], ['MMMCMXCIX', '3999', False, None], ['CDXLIV', '444', False, None]], 99: [['4\n기차\n차표\n표범\n범인', '성공', True, None], ['4\n우유\n유리\n리우\n우유', '4번째 단어에서 실패', True, None], ['1\n사과', '성공', False, None], ['3\n사과\n과자\n사과', '3번째 단어에서 실패', False, None], ['2\n나무\n무지개', '성공', False, None], ['3\n바다\n다리\n리본', '성공', False, None], ['2\napple\nelephant', '성공', False, None]], 100: [['6\n11-01 월급 2500000\n11-02 식비 -12000\n11-02 교통 -1450\n11-05 식비 -30000\n11-05 도서 -18000\n11-07 식비 -9000', '교통 -1,450\n도서 -18,000\n식비 -51,000\n월급 2,500,000\n잔액 2,429,550\n최대 지출일 11-05 (48,000)', True, None], ['1\n01-01 용돈 10000', '용돈 10,000\n잔액 10,000\n최대 지출일 없음', False, None], ['3\n12-31 식비 -5000\n01-01 식비 -5000\n06-15 선물 -5000', '선물 -5,000\n식비 -10,000\n잔액 -15,000\n최대 지출일 01-01 (5,000)', False, None], ['4\n03-02 환불 3000\n03-01 식비 -3000\n03-02 식비 -2000\n03-03 교통 -100000000', '교통 -100,000,000\n식비 -5,000\n환불 3,000\n잔액 -100,002,000\n최대 지출일 03-03 (100,000,000)', False, None]]}
TL = {96: 2, 97: 2, 98: 2, 99: 2, 100: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
