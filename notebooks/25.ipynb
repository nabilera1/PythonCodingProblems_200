> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {121: [['5\n2 4 -10 4 -9', '2 3 0 3 1', True, None], ['1\n1000000000', '0', False, None], ['4\n7 7 7 7', '0 0 0 0', False, None], ['5\n5 4 3 2 1', '4 3 2 1 0', False, None], [None, 'sha256:9e54cd43f5873254d518bf53a375a2c4b8232b87adfcc52d166fba22151385c8', False, 'import random\nrandom.seed(18)\nn = 100000\nprint(n)\nprint(*[random.randint(-10**9, 10**9) for _ in range(n)])']], 122: [['5\n3 1 2 5 4', '3', True, None], ['1\n7', '0', False, None], ['5\n1 2 3 4 5', '0', False, None], ['5\n5 4 3 2 1', '10', False, None], ['4\n2 2 1 1', '4', False, None], [None, '243953', False, 'import random\nrandom.seed(19)\nn = 1000\nprint(n)\nprint(*[random.randint(1, 100) for _ in range(n)])']], 123: [['5\n10 3 7 21 8', '1', True, None], ['2\n-1000000000 1000000000', '2000000000', False, None], ['3\n5 5 9', '0', False, None], [None, '1', False, 'import random\nrandom.seed(20)\nn = 100000\nprint(n)\nprint(*random.sample(range(-10**9, 10**9), n))']], 124: [['5\n3 0 6 1 5', '3', True, None], ['3\n100 100 100', '3', True, None], ['1\n0', '0', False, None], ['1\n7', '1', False, None], ['4\n0 0 0 0', '0', False, None], ['6\n1 1 1 1 1 1', '1', False, None], [None, '66644', False, 'import random\nrandom.seed(21)\nn = 100000\nprint(n)\nprint(*[random.randint(0, 200000) for _ in range(n)])']], 125: [['5\n3 30 34 5 9', '9534330', True, None], ['3\n0 0 0', '0', True, None], ['1\n0', '0', False, None], ['1\n42', '42', False, None], ['2\n12 121', '12121', False, None], ['2\n121 12', '12121', False, None], ['4\n824 8247 938 1399', '93882482471399', False, None], ['3\n10 2 0', '2100', False, None]]}
TL = {121: 2, 122: 2, 123: 2, 124: 2, 125: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
