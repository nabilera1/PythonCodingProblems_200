````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {171: [['4', '7', True, None], ['1', '1', True, None], ['2', '2', False, None], ['3', '4', False, None], ['10', '274', False, None], ['1000', '509672692', False, None], ['1000000', '746580045', False, None]], 172: [['9\n-2 1 -3 4 -1 2 1 -5 4', '6', True, None], ['3\n-5 -2 -9', '-2', True, None], ['1\n-1000000000', '-1000000000', False, None], ['1\n7', '7', False, None], ['5\n1 2 3 4 5', '15', False, None], [None, '320871689762', False, 'import random\nrandom.seed(128)\nn = 200000\nprint(n)\nprint(*[random.randint(-10**9, 10**9) for _ in range(n)])'], [None, '-6771', False, 'import random\nrandom.seed(129)\nn = 200000\nprint(n)\nprint(*[random.randint(-10**9, -1) for _ in range(n)])']], 173: [['8\n10 9 2 5 3 7 101 18', '4', True, None], ['4\n7 7 7 7', '1', True, None], ['1\n5', '1', False, None], ['5\n1 2 3 4 5', '5', False, None], ['5\n5 4 3 2 1', '1', False, None], [None, '890', False, 'import random\nrandom.seed(130)\nn = 200000\nprint(n)\nprint(*[random.randint(-10**9, 10**9) for _ in range(n)])'], [None, '200000', False, 'import random\nrandom.seed(131)\nn = 200000\nprint(n)\nprint(*range(1, n + 1))']], 174: [['4 7\n6 13\n4 8\n3 6\n5 12', '14', True, None], ['1 1\n2 100', '0', False, None], ['1 5\n5 7', '7', False, None], ['3 10\n5 10\n5 10\n5 10', '20', False, None], [None, '31533100824', False, 'import random\nrandom.seed(132)\nn, w = 100, 100000\nprint(n, w)\nfor _ in range(n):\n    print(random.randint(1, 10**5 // 10), random.randint(0, 10**9))']], 175: [['ACAYKP\nCAPCAK', '4', True, None], ['A\nA', '1', False, None], ['A\nB', '0', False, None], ['ABCDEF\nFEDCBA', '1', False, None], [None, '1287', False, "import random\nrandom.seed(133)\nprint(''.join(random.choice('ACGT') for _ in range(2000)))\nprint(''.join(random.choice('ACGT') for _ in range(2000)))"], [None, '2000', False, "import random\nrandom.seed(134)\nprint('A' * 2000)\nprint('A' * 2000)"]]}
TL = {171: 2, 172: 2, 173: 2, 174: 3, 175: 4}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
