````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {81: [['10', '1 9 25 49 81', True, None], ['1', '1', False, None], ['2', '1', False, None], ['15', '1 9 25 49 81 121 169 225', False, None], ['1000', '1 9 25 49 81 121 169 225 289 361 441 529 625 729 841 961 1089 1225 1369 1521 1681 1849 2025 2209 2401 2601 2809 3025 3249 3481 3721 3969 4225 4489 4761 5041 5329 5625 5929 6241 6561 6889 7225 7569 7921 8281 8649 9025 9409 9801 10201 10609 11025 11449 11881 12321 12769 13225 13689 14161 14641 15129 15625 16129 16641 17161 17689 18225 18769 19321 19881 20449 21025 21609 22201 22801 23409 24025 24649 25281 25921 26569 27225 27889 28561 29241 29929 30625 31329 32041 32761 33489 34225 34969 35721 36481 37249 38025 38809 39601 40401 41209 42025 42849 43681 44521 45369 46225 47089 47961 48841 49729 50625 51529 52441 53361 54289 55225 56169 57121 58081 59049 60025 61009 62001 63001 64009 65025 66049 67081 68121 69169 70225 71289 72361 73441 74529 75625 76729 77841 78961 80089 81225 82369 83521 84681 85849 87025 88209 89401 90601 91809 93025 94249 95481 96721 97969 99225 100489 101761 103041 104329 105625 106929 108241 109561 110889 112225 113569 114921 116281 117649 119025 120409 121801 123201 124609 126025 127449 128881 130321 131769 133225 134689 136161 137641 139129 140625 142129 143641 145161 146689 148225 149769 151321 152881 154449 156025 157609 159201 160801 162409 164025 165649 167281 168921 170569 172225 173889 175561 177241 178929 180625 182329 184041 185761 187489 189225 190969 192721 194481 196249 198025 199809 201601 203401 205209 207025 208849 210681 212521 214369 216225 218089 219961 221841 223729 225625 227529 229441 231361 233289 235225 237169 239121 241081 243049 245025 247009 249001 251001 253009 255025 257049 259081 261121 263169 265225 267289 269361 271441 273529 275625 277729 279841 281961 284089 286225 288369 290521 292681 294849 297025 299209 301401 303601 305809 308025 310249 312481 314721 316969 319225 321489 323761 326041 328329 330625 332929 335241 337561 339889 342225 344569 346921 349281 351649 354025 356409 358801 361201 363609 366025 368449 370881 373321 375769 378225 380689 383161 385641 388129 390625 393129 395641 398161 400689 403225 405769 408321 410881 413449 416025 418609 421201 423801 426409 429025 431649 434281 436921 439569 442225 444889 447561 450241 452929 455625 458329 461041 463761 466489 469225 471969 474721 477481 480249 483025 485809 488601 491401 494209 497025 499849 502681 505521 508369 511225 514089 516961 519841 522729 525625 528529 531441 534361 537289 540225 543169 546121 549081 552049 555025 558009 561001 564001 567009 570025 573049 576081 579121 582169 585225 588289 591361 594441 597529 600625 603729 606841 609961 613089 616225 619369 622521 625681 628849 632025 635209 638401 641601 644809 648025 651249 654481 657721 660969 664225 667489 670761 674041 677329 680625 683929 687241 690561 693889 697225 700569 703921 707281 710649 714025 717409 720801 724201 727609 731025 734449 737881 741321 744769 748225 751689 755161 758641 762129 765625 769129 772641 776161 779689 783225 786769 790321 793881 797449 801025 804609 808201 811801 815409 819025 822649 826281 829921 833569 837225 840889 844561 848241 851929 855625 859329 863041 866761 870489 874225 877969 881721 885481 889249 893025 896809 900601 904401 908209 912025 915849 919681 923521 927369 931225 935089 938961 942841 946729 950625 954529 958441 962361 966289 970225 974169 978121 982081 986049 990025 994009 998001', False, None]], 82: [['2 3\n1 2 3\n4 5 6', '1 4\n2 5\n3 6', True, None], ['1 1\n7', '7', False, None], ['1 4\n1 2 3 4', '1\n2\n3\n4', False, None], ['4 1\n1\n2\n3\n4', '1 2 3 4', False, None], ['3 3\n1 2 3\n4 5 6\n7 8 9', '1 4 7\n2 5 8\n3 6 9', False, None]], 83: [['banana', 'a:1 b:0 n:2', True, None], ['python', 'h:3 n:5 o:4 p:0 t:2 y:1', True, None], ['a', 'a:0', False, None], ['zzz', 'z:0', False, None], ['mississippi', 'i:1 m:0 p:8 s:2', False, None], ['abcdefghijklmnopqrstuvwxyz', 'a:0 b:1 c:2 d:3 e:4 f:5 g:6 h:7 i:8 j:9 k:10 l:11 m:12 n:13 o:14 p:15 q:16 r:17 s:18 t:19 u:20 v:21 w:22 x:23 y:24 z:25', False, None]], 84: [['10', '6 4', True, None], ['1', '1 1', True, None], ['2', '2 2', False, None], ['12', '12 6', False, None], ['100', '60 12', False, None], ['360', '360 24', False, None], ['1000', '840 32', False, None]], 85: [['4\n1 2 3 4', '5', True, None], ['3\n5 5 5', '1', True, None], ['1\n7', '0', False, None], ['2\n-1 1', '1', False, None], ['5\n1 10 100 1000 10000', '10', False, None], [None, '200', False, 'import random\nrandom.seed(10)\nn = 300\nprint(n)\nprint(*[random.randint(-50, 50) for _ in range(n)])']]}
TL = {81: 2, 82: 2, 83: 2, 84: 2, 85: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
