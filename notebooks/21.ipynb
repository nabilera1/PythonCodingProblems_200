4. 직접 입력을 넣어 실행해 보려면 새 셀에 `%run p001.py`처럼 입력합니다.

> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {101: [['10', '4 7', True, None], ['100', '25 97', True, None], ['2', '1 2', False, None], ['3', '2 3', False, None], ['4', '2 3', False, None], ['1000', '168 997', False, None], ['999983', '78498 999983', False, None], ['1000000', '78498 999983', False, None]], 102: [['360', '2^3 * 3^2 * 5', True, None], ['97', '97', True, None], ['2', '2', False, None], ['1024', '2^10', False, None], ['999999999989', '999999999989', False, None], ['1000000000000', '2^12 * 5^12', False, None], ['600851475143', '71 * 839 * 1471 * 6857', False, None], ['999966000289', '999983^2', False, None]], 103: [['28', '6 56', True, None], ['1', '1 1', True, None], ['2', '2 3', False, None], ['12', '6 28', False, None], ['496', '10 992', False, None], ['8128', '14 16256', False, None], ['1000000000000', '169 2499694822171', False, None], ['999999999989', '2 999999999990', False, None], ['963761198400', '6720 5531765944320', False, None]], 104: [['36', '12', True, None], ['1', '1', True, None], ['2', '1', False, None], ['13', '12', False, None], ['1000000007', '1000000006', False, None], ['999999999989', '999999999988', False, None], ['1000000000000', '400000000000', False, None], ['720720', '138240', False, None]], 105: [['2 10 1000', '24', True, None], ['3 200 13', '9', True, None], ['5 0 7', '1', False, None], ['5 3 1', '0', False, None], ['0 0 10', '1', False, None], ['0 5 10', '0', False, None], ['123456789 1000000000000000000 1000000007', '228100152', False, None], ['1000000000000000000 1000000000000000000 999999937', '202779993', False, None], ['2 1000000006 1000000007', '1', False, None]]}
TL = {101: 2, 102: 2, 103: 2, 104: 2, 105: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
