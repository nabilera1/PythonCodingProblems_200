> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {56: [['5\n7 1 3 1000000000 2', '3', True, None], ['1\n-4', '-4', False, None], ['3\n3 2 1', '2', False, None], ['7\n5 5 5 1 9 9 9', '5', False, None], [None, '-1565499', False, 'import random\nrandom.seed(1)\nn = 99999\nprint(n)\nprint(*[random.randint(-10**9, 10**9) for _ in range(n)])']], 57: [['banana kiwi apple fig cherry date', 'fig date kiwi apple banana cherry', True, None], ['a', 'a', False, None], ['bb a ccc', 'a bb ccc', False, None], ['cab abc bca', 'abc bca cab', False, None], ['python is a very elegant language', 'a is very python elegant language', False, None]], 58: [['5\nkim 90\nlee 95\npark 90\nchoi 80\nhan 95', 'han 95\nlee 95\nkim 90', True, None], ['3\na 1\nb 2\nc 3', 'c 3\nb 2\na 1', False, None], ['4\nzed 50\namy 50\nbob 50\ncat 50', 'amy 50\nbob 50\ncat 50', False, None], ['4\nx 0\ny 100\nz 0\nw 99', 'y 100\nw 99\nx 0', False, None]], 59: [['6\n10 20 30 40 50 60', '10 20 30\n40 50 60\n10 30 50\n60 50 40 30 20 10', True, None], ['3\n1 2 3', '1 2 3\n1 2 3\n1 3\n3 2 1', False, None], ['4\n4 3 2 1', '4 3 2\n3 2 1\n4 2\n1 2 3 4', False, None], ['7\n1 1 2 3 5 8 13', '1 1 2\n5 8 13\n1 2 5 13\n13 8 5 3 2 1 1', False, None]], 60: [['3 4\n1 4 9\n2 3 4 10', '1 2 3 4 4 9 10', True, None], ['1 1\n5\n5', '5 5', False, None], ['2 3\n10 20\n1 2 3', '1 2 3 10 20', False, None], ['3 1\n1 2 3\n100', '1 2 3 100', False, None], [None, 'sha256:e97ec683f2f6fa7e81638769699372ac767257813843a3593e66048b417108a0', False, 'import random\nrandom.seed(2)\nn, m = 100000, 100000\nprint(n, m)\nprint(*sorted(random.randint(-10**9, 10**9) for _ in range(n)))\nprint(*sorted(random.randint(-10**9, 10**9) for _ in range(m)))']]}
TL = {56: 2, 57: 2, 58: 2, 59: 2, 60: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
