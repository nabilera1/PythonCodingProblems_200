기초 · 문제 041–045

**사용법**
1. 바로 아래 **🔧 채점기 준비** 셀을 먼저 실행합니다(▶).
2. 각 문제의 `%%writefile` 셀에 코드를 작성하고 실행하면 파일로 저장됩니다.
3. 그 아래 `check(번호)` 셀을 실행하면 채점됩니다.
4. 직접 입력을 넣어 실행해 보려면 새 셀에 `%run p001.py`처럼 입력합니다.

> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {41: [['기러기', '회문', True], ['python', '회문 아님', True], ['level', '회문', False], ['a', '회문', False], ['ab', '회문 아님', False], ['토마토', '회문', False], ['역삼역', '회문', False], ['abccba', '회문', False], ['abcdba', '회문 아님', False]], 42: [['Hello, World!', '3', True], ['rhythm', '0', True], ['AEIOU aeiou', '10', False], ['Python 3.12', '1', False], ['The quick brown fox jumps over the lazy dog', '11', False], ['a', '1', False]], 43: [['portable network graphics', 'PNG', True], ['as soon as possible', 'ASAP', True], ['  united   nations  ', 'UN', False], ['Python', 'P', False], ['frequently asked questions', 'FAQ', False], ['read eval print loop', 'REPL', False]], 44: [['hello world', 'l', True], ['abcABC', 'a', True], ['Zzz... y', 'z', False], ['Mississippi', 'i', False], ['The quick brown fox jumps over the lazy dog', 'o', False], ['e', 'e', False], ['To be, or not to be', 'o', False]], 45: [['student@school.kr', 'student\nschool.kr', True], ['a@b', 'a\nb', False], ['kim.dg_2026@mail.example.com', 'kim.dg_2026\nmail.example.com', False], ['python@python.org', 'python\npython.org', False]]}
import subprocess, sys, pathlib

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def check(pid, time_limit=2.0):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    ok = 0
    for i, (inp, exp, shown) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        try:
            r = subprocess.run([sys.executable, str(path)], input=inp + "\n",
                               capture_output=True, text=True, timeout=time_limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({time_limit}초)"); continue
        got = _norm(r.stdout)
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
