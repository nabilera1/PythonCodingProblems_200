4. 직접 입력을 넣어 실행해 보려면 새 셀에 `%run p001.py`처럼 입력합니다.

> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {51: [['5\n8586 8848 8516 8848 8485', '8848 2', True, None], ['1\n100', '100 1', False, None], ['3\n5 4 3', '5 1', False, None], ['4\n1 2 3 4', '4 4', False, None], ['6\n7 7 7 7 7 7', '7 1', False, None]], 52: [['5\n50 50 70 80 90', '3', True, None], ['4\n10 10 10 100', '1', True, None], ['1\n70', '0', False, None], ['3\n1 2 3', '1', False, None], ['3\n0 0 1', '1', False, None], ['6\n100 100 100 100 100 100', '0', False, None], ['7\n33 33 34 33 33 33 34', '2', False, None]], 53: [['5 2\n1 2 3 4 5', '4 5 1 2 3', True, None], ['3 7\n10 20 30', '30 10 20', True, None], ['1 1000000000\n42', '42', False, None], ['4 0\n1 2 3 4', '1 2 3 4', False, None], ['4 4\n1 2 3 4', '1 2 3 4', False, None], ['6 1000000000\n1 2 3 4 5 6', '3 4 5 6 1 2', False, None]], 54: [['4\n1000 -300 -900 500', '1000 700 -200 300\n-200', True, None], ['1\n-5', '-5\n-5', False, None], ['3\n10 20 30', '10 30 60\n10', False, None], ['5\n-1 -1 -1 -1 -1', '-1 -2 -3 -4 -5\n-5', False, None], ['2\n1000000 -1000000', '1000000 0\n0', False, None]], 55: [['5\n3 9 9 4 1', '4', True, None], ['3\n7 7 7', '없음', True, None], ['1\n5', '없음', False, None], ['2\n1 2', '1', False, None], ['2\n2 1', '1', False, None], ['6\n-5 -1 -1 -3 -1 -5', '-3', False, None], ['4\n1000000000 -1000000000 0 0', '0', False, None]]}
TL = {51: 2, 52: 2, 53: 2, 54: 2, 55: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
