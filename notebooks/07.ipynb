> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {31: [['1071 1029', '21', True, None], ['12 18', '6', True, None], ['17 5', '1', False, None], ['7 7', '7', False, None], ['1 1000000000000000000', '1', False, None], ['1000000000000000000 999999999999999999', '1', False, None], ['832040 514229', '1', False, None]], 32: [['12345', '54321', True, None], ['1200', '21', True, None], ['7', '7', False, None], ['10', '1', False, None], ['1000000000000000000', '1', False, None], ['123454321', '123454321', False, None], ['907', '709', False, None]], 33: [['100 5 200', '15', True, None], ['1000 10 2000', '8', True, None], ['1000 1 1000', '0', False, None], ['1 100 1024', '10', False, None], ['500 3 1000', '24', False, None], ['100 7 1000000', '137', False, None], ['5000 2 1000', '0', False, None]], 34: [['97', '소수', True, None], ['1', '소수도 합성수도 아님', True, None], ['2', '소수', False, None], ['4', '합성수', False, None], ['9', '합성수', False, None], ['1000000007', '소수', False, None], ['999999937', '소수', False, None], ['2000000000', '합성수', False, None], ['1999999973', '소수', False, None]], 35: [['5\n3\n-2\n0', '3 6', True, None], ['0', '0 0', True, None], ['7\n0', '1 7', False, None], ['1\n1\n1\n1\n1\n0', '5 5', False, None], ['-5\n-5\n0', '2 -10', False, None], ['1000000000\n1000000000\n0', '2 2000000000', False, None]]}
TL = {31: 2, 32: 2, 33: 2, 34: 2, 35: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
