````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {61: [['1 2 3 4\n3 4 5', '2 5 2', True, None], ['1 1 1\n1', '1 1 0', True, None], ['1 2\n3 4', '0 4 2', False, None], ['5 6 7\n7 6 5', '3 3 0', False, None], ['-1 0 1\n0', '1 3 2', False, None], ['10 20 30 40 50\n1 2 3', '0 8 5', False, None]], 62: [['6\n3 1 4 1 5 3', '1', True, None], ['3\n1 2 3', '없음', True, None], ['1\n7', '없음', False, None], ['2\n9 9', '9', False, None], ['5\n5 4 3 4 5', '4', False, None], [None, '없음', False, 'import random\nrandom.seed(3)\nn = 100000\nprint(n)\nprint(*random.sample(range(10**9), n))'], [None, '583422002', False, 'import random\nrandom.seed(4)\nn = 100000\nprint(n)\na = random.sample(range(10**9), n - 1)\na.append(a[50000])\nprint(*a)']], 63: [['To be, or not to be: that is the question.', '8', True, None], ['Hello hello HELLO!', '1', False, None], ['a', '1', False, None], ['The cat saw the dog; the dog saw the cat.', '4', False, None], ["Is it? It is! 'It' is.", '2', False, None]], 64: [['5\n3 4\n1 1\n1 -1\n2 2\n3 3', '1 -1\n1 1\n2 2\n3 3\n3 4', True, None], ['1\n0 0', '0 0', False, None], ['3\n0 0\n0 0\n-1 5', '-1 5\n0 0\n0 0', False, None], [None, 'sha256:0155831af2ba00958c54e4b103c7662a3c66f4485e749019edbdd81267fc70a2', False, 'import random\nrandom.seed(5)\nn = 100000\nprint(n)\nfor _ in range(n):\n    print(random.randint(-10**5, 10**5), random.randint(-10**5, 10**5))']], 65: [['3\n3 15\n12 25\n3 15', '있음', True, None], ['2\n1 1\n11 1', '없음', True, None], ['1\n2 29', '없음', False, None], ['4\n1 2\n2 1\n1 12\n12 1', '없음', False, None], [None, '없음', False, 'import random\nrandom.seed(8)\ndays = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]\nds = [(m, d) for m in range(1, 13) for d in range(1, days[m - 1] + 1)]\nrandom.shuffle(ds)\nprint(len(ds))\nfor m, d in ds:\n    print(m, d)'], [None, '있음', False, 'import random\nrandom.seed(9)\ndays = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]\nds = [(m, d) for m in range(1, 13) for d in range(1, days[m - 1] + 1)]\nrandom.shuffle(ds)\nds.append(ds[100])\nprint(len(ds))\nfor m, d in ds:\n    print(m, d)']]}
TL = {61: 2, 62: 2, 63: 2, 64: 2, 65: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
