> 정답과 해설은 위키독스 본문의 접힌 상자에 있습니다. 충분히 고민한 뒤에 펼쳐 보세요.

````python
#@title 🔧 채점기 준비 (먼저 실행하세요)
TESTS = {46: [['3\nVeni, vidi, vici.', 'Yhql, ylgl, ylfl.', True, None], ['-3\nYhql, ylgl, ylfl.', 'Veni, vidi, vici.', True, None], ['1\nxyz XYZ', 'yza YZA', False, None], ['13\nHello, World!', 'Uryyb, Jbeyq!', False, None], ['0\nabc', 'abc', False, None], ['25\nPython 3.12', 'Oxsgnm 3.12', False, None], ['-25\nabc', 'bcd', False, None]], 47: [['the quick brown fox', 'fox brown quick the', True, None], ['hello', 'hello', True, None], ['  many   spaces here  ', 'here spaces many', False, None], ['a b c d e', 'e d c b a', False, None], ['to be or not to be', 'be to not or be to', False, None]], 48: [['3\napple 1200 3\nmilk 2500 1\nbread 4800 2', 'apple          3,600\nmilk           2,500\nbread          9,600\n--------------------\nTOTAL         15,700', True, None], ['1\nx 1 1', 'x                  1\n--------------------\nTOTAL              1', False, None], ['2\nlaptop 1000000 100\npen 500 3', 'laptop    100,000,000\npen            1,500\n--------------------\nTOTAL     100,001,500', False, None]], 49: [['aaabccdddd', 'a3b1c2d4', True, None], ['abc', 'a1b1c1', True, None], ['a', 'a1', False, None], ['zzzzzzzzzzzz', 'z12', False, None], ['aabbaa', 'a2b2a2', False, None], ['', 'a100000', False, None]], 50: [['Py+thon2026', '안전', True, None], ['python2026!', '취약', True, None], ['Ab1!', '취약', False, None], ['PYTHON2026!', '취약', False, None], ['Python!!!!', '취약', False, None], ['Python2026', '취약', False, None], ['Aa1!aaaa', '안전', False, None], ['aB3$eFgH', '안전', False, None]]}
TL = {46: 2, 47: 2, 48: 2, 49: 2, 50: 2}
import subprocess, sys, pathlib, hashlib

_GEN_CACHE = {}

def _norm(s):
    return "\n".join(l.rstrip() for l in s.strip("\n").splitlines()).strip("\n")

def _input_of(inp, gen):
    if gen is None:
        return inp
    if gen not in _GEN_CACHE:
        _GEN_CACHE[gen] = subprocess.run([sys.executable, "-c", gen], capture_output=True, text=True).stdout
    return _GEN_CACHE[gen]

def check(pid):
    """p{번호}.py 를 모든 테스트로 실행해 채점합니다."""
    path = pathlib.Path(f"p{pid:03d}.py")
    if not path.exists():
        print(f"먼저 위의 %%writefile p{pid:03d}.py 셀을 실행하세요."); return
    limit = TL[pid]
    ok = 0
    for i, (inp, exp, shown, gen) in enumerate(TESTS[pid], 1):
        tag = f"예제 {i}" if shown else f"숨은 테스트 {i}"
        data = _input_of(inp, gen)
        try:
            r = subprocess.run([sys.executable, str(path)], input=data.rstrip("\n") + "\n",
                               capture_output=True, text=True, timeout=limit)
        except subprocess.TimeoutExpired:
            print(f"⏱ {tag}: 시간 초과 ({limit}초)"); continue
        got = _norm(r.stdout)
        if exp.startswith("sha256:"):
            got_cmp = "sha256:" + hashlib.sha256(got.encode()).hexdigest()
        else:
            got_cmp = got
        if r.returncode:
            last = r.stderr.strip().splitlines()[-1] if r.stderr.strip() else "알 수 없는 오류"
            print(f"💥 {tag}: 실행 오류 → {last}")
        elif got_cmp == exp:
            ok += 1; print(f"✅ {tag}")
        elif shown:
            print(f"❌ {tag}\n   입력: {inp!r}\n   기대: {exp!r}\n   실제: {got!r}")
        elif gen is not None:
            print(f"❌ {tag}  (큰 입력)")
        else:
            print(f"❌ {tag}  (입력: {inp[:60]!r})")
    total = len(TESTS[pid])
    print("─" * 30)
    print(f"🎉 {pid:03d}번 통과! ({ok}/{total})" if ok == total else f"{ok}/{total} 통과 — 다시 도전해 보세요.")
print('채점기 준비 완료!')
````
